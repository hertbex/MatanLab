### Доказательство

Рассмотрим рекуррентное отображение:

$$
x_{n+1} = r x_n (1 - x_n)
$$

Пусть выполнено условие:

$$
0 < x_0 < 1, \quad 0 < r \le 4.
$$

Докажем по индукции, что при этих условиях для всех $n \in \mathbb{N}$ выполняется:

$$
0 < x_n < 1.
$$

---

#### База индукции

По условию задачи:

$$
0 < x_0 < 1.
$$

---

#### Переход индукции

Предположим, что для некоторого $n$ выполнено:

$$
0 < x_n < 1.
$$

Тогда:

$$
x_{n+1} = r x_n (1 - x_n).
$$

Так как $0 < x_n < 1$, то:

$$
1 - x_n > 0 \quad \Rightarrow \quad x_n(1 - x_n) > 0.
$$

Поскольку также $r > 0$, получаем:

$$
x_{n+1} = r x_n (1 - x_n) > 0.
$$

Теперь докажем верхнюю грань.

Так как $0 < x_n < 1$, то:

$$
x_n(1 - x_n) \le \frac{1}{4},
$$

где максимум функции $x(1-x)$ достигается при $x=\tfrac{1}{2}$.

Так как $r \le 4$, получаем:

$$
x_{n+1} = r x_n (1 - x_n) \le 4 \cdot \frac{1}{4} = 1.
$$

Покажем строгий знак «<»:

Так как $0 < x_n < 1$, то:

$$
x_n(1 - x_n) < \frac{1}{4}.
$$

А так как $r < 4$ или даже при $r = 4$:

$$
x_{n+1} = r x_n (1 - x_n) < 4 \cdot \frac{1}{4} = 1.
$$

Таким образом:

$$
0 < x_{n+1} < 1.
$$

---

#### Вывод

Мы доказали, что если:

$$
0 < x_n < 1,
$$

то обязательно:

$$
0 < x_{n+1} < 1.
$$

А значит по принципу математической индукции:

$$
\forall n \in \mathbb{N},\quad 0 < x_0 < 1 \Rightarrow 0 < x_n < 1.
$$

---

$$
\boxed{\text{Тем самым утверждение доказано.}}
$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Настройки для лучшего отображения
plt.rcParams['figure.figsize'] = (10, 6)

# ============================================
# 1. ЛОГИСТИЧЕСКОЕ ОТОБРАЖЕНИЕ
# ============================================

print("="*60)
print("1. ГРАФИК ЛОГИСТИЧЕСКОГО ОТОБРАЖЕНИЯ")
print("="*60)

x = np.linspace(0, 1, 500)
r_values = [0.5, 1.5, 2.5, 3.5]

plt.figure()
for r in r_values:
    y = r * x * (1 - x)
    plt.plot(x, y, label=f'r = {r}', linewidth=2)

plt.plot(x, x, 'k--', label='y = x', alpha=0.5)
plt.xlabel('$x_{n-1}$')
plt.ylabel('$x_n$')
plt.title('Логистическое отображение: $x_n = r x_{n-1} (1 - x_{n-1})$')
plt.legend()
plt.grid(True)
plt.xlim(0, 1)
plt.ylim(0, 1.1)
plt.show()

# ============================================
# 2. МОЙ ВАРИАНТ (N=1)
# ============================================

print("\n" + "="*60)
print("2. ГРАФИК МОЕГО ВАРИАНТА (N=1)")
print("="*60)
print("Формула: $x_n = r x_{n-1} (1 - x_{n-1}^2)$")
print(f"Диапазон r: [0, {3*np.sqrt(3)/2:.3f}]")

x = np.linspace(0, 1, 500)
r_values_var = [0.5, 1.0, 1.5, 2.0, 2.5]

# Фильтруем значения, не превышающие максимальное
r_max = 3 * np.sqrt(3) / 2
r_values_var = [r for r in r_values_var if r <= r_max]

plt.figure()
for r in r_values_var:
    y = r * x * (1 - x**2)
    plt.plot(x, y, label=f'r = {r}', linewidth=2)

plt.plot(x, x, 'k--', label='y = x', alpha=0.5)
plt.xlabel('$x_{n-1}$')
plt.ylabel('$x_n$')
plt.title('Мой вариант (N=1): $x_n = r x_{n-1} (1 - x_{n-1}^2)$')
plt.legend()
plt.grid(True)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.show()

# ============================================
# 3. СРАВНЕНИЕ ПРИ r = 2.0
# ============================================

print("\n" + "="*60)
print("3. СРАВНИТЕЛЬНЫЙ ГРАФИК ПРИ r = 2.0")
print("="*60)

r = 2.0
x = np.linspace(0, 1, 500)
y_logistic = r * x * (1 - x)
y_variant = r * x * (1 - x**2)

plt.figure()
plt.plot(x, y_logistic, 'b-', label='Логистическое', linewidth=2.5)
plt.plot(x, y_variant, 'r-', label='Мой вариант (N=1)', linewidth=2.5)
plt.plot(x, x, 'k--', label='y = x', alpha=0.5)
plt.xlabel('$x_{n-1}$')
plt.ylabel('$x_n$')
plt.title(f'Сравнение при $r = {r}$')
plt.legend()
plt.grid(True)
plt.xlim(0, 1)
plt.ylim(0, 1)
plt.show()
```

### Вывод: как параметр $r$ влияет на поведение функции

Для логистического отображения $x_n = r x_{n-1} (1 - x_{n-1})$:

1. **При $r \in [0, 1]$**: функция полностью лежит ниже линии $y = x$, что означает, что все траектории стремятся к 0.

2. **При $r \in (1, 3]$**: появляется вторая неподвижная точка $x^* = 1 - \frac{1}{r}$, которая является устойчивой.

3. **При $r > 3$**: начинаются бифуркации — неподвижная точка теряет устойчивость, возникают периодические орбиты.

4. **При $r \approx 3.57$ и выше**: система демонстрирует хаотическое поведение.

5. **При $r = 4$**: максимальный хаос.

Параметр $r$ определяет:
- Высоту параболы (максимальное значение функции)
- Количество и положение неподвижных точек
- Динамический режим системы (устойчивость, периодичность, хаос)


### Сравнение логистического отображения и моего варианта (N=1)

#### Сходства:
1. **Общая форма**: обе функции имеют колоколообразную форму на отрезке $[0, 1]$
2. **Корни**: обе имеют корни в точках $x = 0$ и $x = 1$
3. **Динамические режимы**: обе демонстрируют схожие режимы:
   - При малых $r$: сходимость к 0
   - При средних $r$: устойчивые неподвижные точки
   - При больших $r$: сложная динамика
4. **Биологическая интерпретация**: обе моделируют рост популяции с насыщением

#### Различия:
1. **Математическая форма**:
   - Логистическое: $rx(1-x)$ (квадратичная функция)
   - Мой вариант: $rx(1-x^2)$ (кубическая функция)

2. **Положение максимума**:
   - Логистическое: $x_{\text{max}} = 0.5$, $f_{\text{max}} = r/4$
   - Мой вариант: $x_{\text{max}} = 1/\sqrt{3} \approx 0.577$, $f_{\text{max}} = 2r/(3\sqrt{3}) \approx 0.385r$

3. **Диапазон параметра $r$**:
   - Логистическое: $r \in [0, 4]$ для $f(x) \leq 1$
   - Мой вариант: $r \in [0, 3\sqrt{3}/2 \approx 2.598]$ для $g(x) \leq 1$

4. **Симметрия**:
   - Логистическое симметрично относительно $x = 0.5$
   - Мой вариант несимметричен

#### Причины сходств и различий:

**Сходства** вызваны общей концепцией моделирования:
- Обе модели описывают рост с ограничениями
- Обе учитывают внутривидовую конкуренцию
- Обе имеют S-образную кривую роста

**Различия** вызваны:
1. **Разной степенью нелинейности**: линейная по $x$ конкуренция в логистическом vs квадратичная в моем варианте
2. **Разной моделью взаимодействия**: мой вариант предполагает более сильное подавление роста при высокой численности
3. **Разной чувствительностью к перенаселению**

**Биологическая интерпретация различий**:
- Логистическое отображение: конкуренция пропорциональна численности
- Мой вариант: конкуренция растет быстрее численности (квадратично), что может соответствовать видам с территориальным поведением или агрегациями
